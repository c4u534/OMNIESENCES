````
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Known Mersenne exponents (p) for 2^p - 1
# We use the first 12 for high-precision local calculation
mersenne_p = [2, 3, 5, 7, 13, 17, 19, 31, 61, 89, 107, 127]
mersenne_values = [2**p - 1 for p in mersenne_p]

# Derivational Indexing: Calculating the 'Congruency Ratio'
# We hypothesize a ratio: R = (log(Mp)^3 / log(Mp)^2) / p normalized by a geometric factor
derivation_data = []
for i, p in enumerate(mersenne_p):
    Mp = mersenne_values[i]
    ln_Mp = np.log(float(Mp))

    # Ratio of cubed to squared factors (logarithmic reduction)
    ratio_raw = ln_Mp

    # Searching for the Pi-reduction factor: R = ratio / (p * C)
    # We test the hypothesis that the relationship between the exponent p
    # and the logarithmic density approaches a pi-congruent state.
    pi_approx = (ln_Mp / p)

    derivation_data.append({
        "p": p,
        "log_Mp": ln_Mp,
        "Reduction_Ratio": pi_approx,
        "Pi_Error": abs(pi_approx - np.pi) / np.pi
    })

mersenne_analysis_df = pd.DataFrame(derivation_data)
display(mersenne_analysis_df)

# Visualization of the Convergence toward the Congruency Ratio
plt.figure(figsize=(10, 5))
plt.plot(mersenne_analysis_df['p'], mersenne_analysis_df['Reduction_Ratio'], marker='o', linestyle='-', color='purple')
plt.axhline(y=np.log(2), color='red', linestyle='--', label='ln(2) Boundary')
plt.title("Mersenne Derivational Index: Convergence Analysis")
plt.xlabel("Mersenne Exponent (p)")
plt.ylabel("Congruency Ratio (ln(Mp)/p)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````
